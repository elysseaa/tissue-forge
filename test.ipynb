```python
import tissue_forge as tf
import ipywidgets as ipw
from IPython.display import display
import importlib
import particleModifier
importlib.reload(particleModifier)

from particleModifier import (
    set_ptype_color_text,
    set_ptype_mass_text,
    set_ptype_number_text,
    set_ptype_radius_text,
    batch_edit_ptypes,
    add_simulation_controls,
    add_export_import_buttons,
    feedback_output
)

# === Initialize Tissue Forge ===
try:
    tf.init()
except RuntimeError as e:
    print(f"Tissue Forge already initialized: {e}")

# === Create multiple test ParticleTypes ===
ptypes = []
for i in range(3):
    pt = tf.ParticleType()
    pt.name = f"TestType{i+1}"
    pt.mass = 1.0 + i
    pt.radius = 1.0 + i * 0.2
    pt.style.color = tf.FVector3(0.3 + i*0.2, 0.3, 0.3)
    pt.factory(nr_parts=5 * (i + 1))
    for j, p in enumerate(pt.parts):
        p.position = tf.FVector3(j * 1.5, i * 2.0, 0)
    ptypes.append(pt)

# === Display Title ===
display(ipw.HTML("<h3>🎛️ Particle Type Property Widgets</h3>"))

# === Show individual property editors for the first ParticleType ===
color_text_box, *_ = set_ptype_color_text(ptypes[0])
mass_text_box, _ = set_ptype_mass_text(ptypes[0])
number_text_box, _ = set_ptype_number_text(ptypes[0])
radius_text_box, _ = set_ptype_radius_text(ptypes[0])

display(ipw.HTML("<b>Single Particle Type Editor</b>"))
display(ipw.VBox([
    color_text_box,
    mass_text_box,
    number_text_box,
    radius_text_box
]))

# === Show batch editor dropdown (uses dropdown inside the function now) ===
display(ipw.HTML("<b>Batch Editor</b>"))
batch_edit_ptypes(ptypes)

# === Show simulation controls ===
display(ipw.HTML("<b>Simulation Controls</b>"))
add_simulation_controls()

# === Show export/import buttons ===
display(ipw.HTML("<b>Configuration Save/Load</b>"))
add_export_import_buttons(ptypes)

# === Show feedback output console ===
display(ipw.HTML("<b>Live Feedback</b>"))
display(feedback_output)
```
