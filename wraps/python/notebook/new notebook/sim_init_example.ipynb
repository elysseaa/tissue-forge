# Simulation Initialization

```python
import tissue_forge as tf
tf.notebook.init()
```

Running the previous cell should display a widget interface for customizing simulation initialization. It consists of three tabs:
* Settings (contains basic settings like cutoff, dim, and dt params)
* Boundary conditions (contains bc param)
* Advanced (contains all other params)

**Checkbox:** Each customizable parameter has a checkbox associated with it. If it is unchecked, tf.init() will be run with default values. If it is checked, clicking "Initialize" will read the data from the widget and run tf.init() with the data.

### Boundary Conditions
This tab contains customization options for the bc param. It is formatted a bit differently than the settings and advanced tab. 
* **Apply Everywhere:** If this checkbox is checked, the bc param will be initialized with the homogeneous condition selected in the dropdown to the right of it. If it isn't checked, the bc param will be initialized to the heterogeneous conditions set in the tabs below it.
* **Heterogeneous Conditions:** Users can customize heterogeneous conditions for 6 faces. They must press the "Save Settings" button for the conditions to be saved.

### Tips
* **Feedback:** The box at the bottom is an output box which will display feedback messages (like "Successfully Initialized!"). Any errors that occur upon initialization will also be displayed here.
* **Pressing initialize twice:** If the user tries to press the initialize button twice without restarting the kernel, the kernel will crash. The user will see the statement "Initializing" in the output box, which is confirmation that the program tried to run tf.init(). This is expected behavior, and the user must restart the kernel to reinitialize the simulation.
